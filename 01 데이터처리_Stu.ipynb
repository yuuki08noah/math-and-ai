```python
import numpy as np
a = np.array([[2,-1],[1,3]])
if np.linalg.det(a) == 0 :
    print('역행렬이 존재하지 않습니다.')
else :
    print('행렬 a의 역행렬: \n', np.linalg.inv(a))
```

**P.31 [코딩으로 해결하기]**

```python
import numpy as np
a = np.array([[13,15],[13.5,20]])
b = np.array([13.8,16.1])
x = np.linalg.inv(a)@b # linalg.inv: 역행렬
print('x : %.2f, y : %.2f'%(x[0], x[1]))
```
